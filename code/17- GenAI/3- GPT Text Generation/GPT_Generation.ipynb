# GenAI Lab 3: GPT Text Generation

### Goal: Load pre-trained GPT-2 and generate text autoregressively using Greedy Search vs Sampling strategies (Temperature, Top-p).


```python
# Step 1) استيراد مكتبات توليد النصوص / Import GPT-2 model and tokenizer
from transformers import GPT2LMHeadModel, GPT2Tokenizer
```

```python
# Step 2) تحميل النموذج والتوكن الحروف / Load model and tokenizer from pretrained cache
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
```

```python
# Step 3) تهيئة المدخلات النصية / Encode prompt text
prompt = "Artificial Intelligence is going to revolutionize"
inputs = tokenizer.encode(prompt, return_tensors="pt")
print("Encoded Input ids:", inputs)
```

```python
# Step 4) التوليد بطريقة البحث الجشع / Greedy Search Generation
outputs_greedy = model.generate(inputs, max_length=40)
greedy_text = tokenizer.decode(outputs_greedy[0], skip_special_tokens=True)
print("Greedy Search Output:")
print(greedy_text)
```

```python
# Step 5) التوليد بطريقة أخذ العينات العشوائية مع تحسينات / Sampling Generation (Temp, Top-p)
outputs_sampling = model.generate(
    inputs, 
    max_length=40, 
    do_sample=True, 
    temperature=0.8, 
    top_k=50, 
    top_p=0.92
)
sampling_text = tokenizer.decode(outputs_sampling[0], skip_special_tokens=True)
print("Sampling Generation (Temp=0.8, Top-p=0.92) Output:")
print(sampling_text)
```
