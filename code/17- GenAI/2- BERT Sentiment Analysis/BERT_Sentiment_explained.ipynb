# GenAI Lab 2: BERT Sentiment Analysis

### الهدف: استدعاء نموذج مصنف لغوي قوي من عائلة BERT باستخدام مكتبة Hugging Face لتصنيف مشاعر الجمل بدقة عالية وبأقل سطر برمجية.


### أولاً: استيراد أدوات مكتبة Hugging Face (Import Transformers)

نقوم باستيراد دالة `pipeline` التي تختصر خطوات التقطيع، التحميل، والتنبؤ في كائن واحد.


```python
# Step 1) تثبيت واستيراد مكتبة Hugging Face / Install and import Transformers
# %pip install transformers -q
from transformers import pipeline
```

### ثانياً: تحميل نموذج BERT المخصص للمشاعر (Initialize Model Pipeline)

نقوم بتحميل نسخة DistilBERT المدربة مسبقاً على تصنيف مشاعر الجمل الإنجليزية.


```python
# Step 2) بناء خط معالجة لتصنيف المشاعر / Initialize sentiment classifier pipeline
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
```

### ثالثاً: التنبؤ وتفسير النتائج (Predict and Evaluate)

نقوم بإدخال الجمل واستخراج النتيجة (إيجابي/سلبي) مع نسبة ثقة النموذج بالتوقع.


```python
# Step 3) التنبؤ بمشاعر الجمل / Make predictions
sentences = [
    "I absolutely love learning about Generative AI and Transformers!",
    "This tutorial is quite confusing and hard to follow.",
    "The weather is okay today."
]

print("BERT Sentiment Classifier results:")
print("="*50)
for s in sentences:
    res = classifier(s)[0]
    print(f"Text: '{s}'")
    print(f"Label: {res['label']} | Score: {res['score']:.4f}\n")
```
