# NLP — Text Cleaning

## الترتيب / Flow
1. استيراد المكتبات - Import libraries
2. Regex أساسي - Basic regex patterns
3. دالة تنظيف النص - `clean_text()` pipeline
4. تطبيع الأحرف - Lowercase normalization
5. تطبيق على نصوص حقيقية - Apply to sample texts

```python
# Step 1) استيراد المكتبات / Import libraries
# pip install nltk -q
import re
import html
import unicodedata
```

```python
# Step 2) Regex أساسي / Basic regex
sample = "Contact: info@company.com or visit https://example.com/article"

print("Emails:", re.findall(r'\S+@\S+', sample))
print("URLs:", re.findall(r'https?://\S+', sample))
print("Words:", re.findall(r'\b[a-zA-Z]+\b', sample))
```

```python
# Step 3) دالة تنظيف النص / Text cleaning pipeline
def clean_text(text: str) -> str:
    text = html.unescape(text)
    text = re.sub(r'<[^>]+>', '', text)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    text = re.sub(r'\+?[\d\-\(\)\s]{9,}', '', text)
    text = re.sub(r'@\w+|#\w+', '', text)
    text = text.encode('ascii', 'ignore').decode('ascii')
    text = re.sub(r'[^\w\s\.\!\?]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

raw = '<p>John said: "AI is amazing!!!!" Visit: https://ai.com @john #AI</p>'
print("Before:", raw)
print("After: ", clean_text(raw))
```

```python
# Step 4) تطبيع الأحرف / Lowercase normalization
def normalize_case(text: str) -> str:
    return text.lower()

examples = [
    "Natural Language Processing",
    "ARTIFICIAL INTELLIGENCE",
    "iPhone vs Android",
]

for ex in examples:
    print(f"{ex:<35} -> {normalize_case(ex)}")
```

```python
# Step 5) تطبيق على نصوص حقيقية / Apply pipeline
samples = [
    '<p>Best #NLP course! Email: admin@site.com https://x.com/post</p>',
    'Posted: 2024-01-15 | @tech_blog Read more: www.example.com/article',
]

for raw in samples:
    cleaned = normalize_case(clean_text(raw))
    print(f"RAW:     {raw}")
    print(f"CLEANED: {cleaned}\n")
```
