```python
# Cell 1: Import required libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
# Cell 2: Load dataset
dataset = pd.read_csv('Position_Salaries.csv')
dataset
```

```python
# Cell 3: Select feature and target
X = dataset.iloc[:, 1:2].values
y = dataset.iloc[:, 2].values
```

```python
# Cell 4: Train Random Forest Regression model
regressor = RandomForestRegressor(n_estimators=10, random_state=0)
regressor.fit(X, y)
```

```python
# Cell 5: Predict all training values for evaluation
y_pred_all = regressor.predict(X)
```

```python
# Cell 6: Calculate Performance Metrics (R^2, MSE, MAE)
r2 = r2_score(y, y_pred_all)
mse = mean_squared_error(y, y_pred_all)
mae = mean_absolute_error(y, y_pred_all)

print(f"R-squared (R^2) Score: {r2:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
```

```python
# Cell 7: Predict salary at level 6.5
y_pred = regressor.predict(np.array([[6.5]]))
print(f"Predicted salary for level 6.5: {y_pred[0]:.2f}")
```

```python
# Cell 8: Visualize Random Forest results (Higher Resolution & Smooth Steps)
X_grid = np.arange(X.min(), X.max(), 0.01).reshape(-1, 1)
plt.scatter(X, y, color='red', label='Actual Data')
plt.plot(X_grid, regressor.predict(X_grid), color='blue', label='Random Forest Model')
plt.title('Truth or Bluff (Random Forest Regression)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.legend()
plt.show()
```
